```python
from pyspark.sql.functions import current_timestamp

df_stream = spark.readStream.format("cloudFiles").option("cloudFiles.format", "csv").option("header", "true").option("inferSchema", "true").option("cloudFiles.schemaLocation", "/Volumes/retail_project_b3/bronze/_checkpoints/geolocation_schema").load("/Volumes/retail_project_b3/landing/raw_data/olist_geolocation_dataset.csv").withColumn("ingest_ts", current_timestamp())

query = df_stream.writeStream.format("delta").option("checkpointLocation", "/Volumes/retail_project_b3/bronze/_checkpoints/geolocation").outputMode("append").trigger(availableNow=True).start("/Volumes/retail_project_b3/bronze/geolocation/")
query.awaitTermination()
print(f"✅ Processed {query.lastProgress.get('numInputRows', 0) if query.lastProgress else 0} records")
```
